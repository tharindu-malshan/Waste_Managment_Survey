# **Analyze disposal method**

```python
df["Q21. Disposal Method"].value_counts()
```

```python
df["Q21. Disposal Method"].value_counts(normalize=True).mul(100).round(2)
```

```python
counts = df["Q21. Disposal Method"].value_counts()

short_names = []

for value in counts.index:

    if "Municipal collection service" in value:
        municipal = "Municipal"
    else:
        municipal = ""

    if "Private collection service" in value:
        private = "Private"
    else:
        private = ""

    if "Composting" in value:
        compost = "Compost"
    else:
        compost = ""

    if "Burning in an open area" in value:
        burning = "Burning"
    else:
        burning = ""

    if "Burying" in value:
        burying = "Burying"
    else:
        burying = ""

    if "Dumping in open areas" in value:
        dumping = "Dumping"
    else:
        dumping = ""

    if "By the road or street side" in value:
        road = "Roadside"
    else:
        road = ""

    if "Hand over to municipal collection services" in value:
        handover = "Hand Over"
    else:
        handover = ""

    short = ", ".join(
        x for x in [
            municipal, private, compost,
            burning, burying, dumping,
            road, handover
        ] if x
    )

    short_names.append(short)

counts.index = short_names

counts.plot(kind="bar")

plt.title("Disposal Methods")
plt.xlabel("Method")
plt.ylabel("Respondents")
plt.xticks(rotation=90)
plt.show()
```

# **Analyze numerical waste variables**

**Bags**

```python
bag_columns = [
    "Q12a. Food Waste (bags)",
    "Q12b. Garden Waste (bags)",
    "Q12c. Plastic (bags)",
    "Q12d. Paper (bags)",
    "Q12e. Glass (bags)",
    "Q12f. Metal (bags)",
    "Q12g. E-Waste (bags)",
    "Q12h. Textile (bags)",
    "Q12i. Rubber (bags)",
    "Q12j. Coconut (bags)",
    "Q12k. Packaging (bags)",
    "Q12l. Medical (bags)",
    "Q12m. Other (bags)"
]

df[bag_columns].describe()
```

```python
for col in bag_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

average_waste = df[bag_columns].mean()

print(average_waste.round(2))
```

```python
average_waste.plot(kind="bar")

plt.title("Average Waste Generated by Type")
plt.xlabel("Waste Type")
plt.ylabel("Average Number of Bags")
plt.xticks(rotation=90)
plt.show()
```

**Bins**
